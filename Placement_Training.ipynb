```python
def monkey_bal():
  try:
    n=int(input())
    k=int(input())
    j=int(input())
    m=int(input())
    p=int(input())
    if n<0 or k<=0 or j<=0 or m<0 or p<0:
      print("INVALID INPUT")
      return

    ate_bananas = m/k
    ate_peanuts = p/j
    total_ate = ate_bananas + ate_peanuts

    if (m % k) != 0 or (p % j) != 0:
      total_ate+=1

    monkeys_bal = n - total_ate
    if monkeys_bal < 0:
      monkeys_bal = 0

    print(f"number of monkeys left on the tree:{monkeys_bal}")

  except ValueError:
    print("INVALID INPUT")

if __name__ == "__main__":
  monkey_bal()
```
